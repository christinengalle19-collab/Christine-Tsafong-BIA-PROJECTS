```python
#Load the clean dataset
df = pd.read_csv('df_cleaned.csv')
```

```python
#displays first rows
df.head()
df.info()
```

```python
#Handling missing values
df.interpolate(method='linear', inplace=True)
```

```python
#Detect and Handle Outliers
#Using IQR Method
Q1 = df['close'].quantile(0.25)
Q3 = df['close'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

#outliers = df[(df['Close'] < lower) | (df['Close'] > upper)]
#Handle Outliers
df = df[(df['close'] >= lower) & (df['close'] <= upper)]



```

```python
#Error Detection
#Check for unrealistic values:
df.describe()

#checks
df[df['close'] < 0]
df[df['volume'] < 0]


#fix
df = df[df['close'] > 0]
df = df[df['volume'] > 0]
```

```python
# Data Transformation
# Feature Engineering

#Rolling Average
df['MA_10'] = df['close'].rolling(window=10).mean()
df['MA_30'] = df['close'].rolling(window=30).mean()


#Volatility (Standard Deviation)
df['volatility'] = df['close'].rolling(window=10).std()
#Daily Returns
df['Returns'] = df['close'].pct_change
```

```python

#Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['open','high','low','close','volume']] = scaler.fit_transform(
    df[['open','high','low','close','volume']]
)
```

```python
import pandas as pd
#Encoding Categorical Variables
df = pd.get_dummies(df, columns=['ticker'], drop_first=True)
```

##Because this cell had already been executed, the column was removed during a previous operation. Therefore, it is no longer present in the dataset

```python
#Integration and Final Dataset
df.dropna(inplace=True)
df.head()
```

```python
#Data Splitting
#Time Series Split
from sklearn.model_selection import TimeSeriesSplit

tscv = TimeSeriesSplit(n_splits=5)
for train_index, test_index in tscv.split(df):
    train = df.iloc[train_index]
    val = df.iloc[test_index]
    test = val.iloc[int(len(val) * 0.5):]
    val = val.iloc[:int(len(val) * 0.5)]
    print(train.shape, val.shape, test.shape)
```

```python
# Save Data
df.to_csv('final_cleaned_data.csv')
train.to_csv('train.csv')
val.to_csv('val.csv')
test.to_csv('test.csv')
```
